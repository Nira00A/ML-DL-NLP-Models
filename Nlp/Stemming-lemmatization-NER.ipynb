```python
## Tokenizatin : Tokens are formed
import nltk
nltk.download('punkt_tab')

#Word Tokenization
from nltk.tokenize import word_tokenize
```

```python
text = "He stepped onto the broadway and was hit by a car. \nHe was rushed to the hospital and is in critical condition."
tokens = word_tokenize(text)
print("Original Text: ", text)
print("Tokens: ", tokens)
```

```python
#Sentence Tokenization

from nltk.tokenize import sent_tokenize
sentences = sent_tokenize(text)
print("Sentences: ", sentences)
```

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
stop_words = set(stopwords.words('English'))
print("Stop Words: ", stop_words)
filtered_words =[w for w in tokens if w not in stop_words]

print("Filtered Words: ", filtered_words)
```

## Stemming

Converts the words to their root form. For example, the words "running", "runs", and "ran" would all be reduced to the root word "run". The main purpose of stemming is to reduce inflectional forms and sometimes derivationally related forms of a word to a common base form.

## Lemmatization

Converts the words to their base or dictionary form, known as a lemma. Unlike stemming, lemmatization considers the context and the meaning of the word, which allows it to produce more accurate results. For example, the word "better" would be lemmatized to "good", while "running" would be lemmatized to "run".

```python
from nltk.stem import PorterStemmer
words = ["Organize", "Organizes", "Organizing", "factual" , "running"]
ps = PorterStemmer()

stem_1 = [ps.stem(word) for word in words]
print("Stemmed Words: ", stem_1)
```

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()
lemmatized_words = [ps.stem(word) for word in filtered_words]
print("Filtered Words: ", filtered_words)
print("Lemmatized Words: ", lemmatized_words)
```

```python
from nltk.stem import WordNetLemmatizer
wn = WordNetLemmatizer()
nltk.download('wordnet')
lemmatized_words = [wn.lemmatize(word) for word in filtered_words]
print("Filtered Words: ", filtered_words)
print("Lemmatized Words: ", lemmatized_words)
```

```python
def preprocess_text(text):
    # Tokenization
    words = word_tokenize(text)

    # Stop Words Removal
    stop_words = set(stopwords.words('english'))
    filtered_words = [w for w in words if w.lower() not in stop_words]

    # Stemming
    ps = PorterStemmer()
    stem_words = [ps.stem(w) for w in filtered_words]

    # Joining them all
    sentence = " ".join(stem_words)

    print("The text given by the user:", text)
    print("The generated text:", sentence)

preprocess_text("The day I will earn , I will gift goddies to my loved ones.")
```
