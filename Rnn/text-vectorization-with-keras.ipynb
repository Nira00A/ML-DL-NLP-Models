## Text Token Generattion and Padding

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
```

```python
from tensorflow.keras.layers import TextVectorization

text_vectorization = TextVectorization()
```

```python
text_vectorization.adapt(["I love machine learning", "Deep learning is great", "Recurrent neural networks are powerful"])
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
# Initialize the Tokenizer with an out-of-vocabulary token

tokenizer = Tokenizer(oov_token="<OOV>")
```

```python
docs = ["I love machine learning", "Deep learning is great", "Recurrent neural networks are powerful", "Unknown words here"]
```

```python
tokenizer.fit_on_texts(docs)
```

```python
tokenizer.word_index
```

```python
# Sentence placement with their indexes

sequence = tokenizer.texts_to_sequences(docs)
sequence
```

```python
# For padding the sequences to make them of the same length
from tensorflow.keras.preprocessing.sequence import pad_sequences

sequence_padded = pad_sequences(sequence, padding="post")
sequence_padded
```
