## Tensrflow Math

```python
x_abs = tf.constant([-2.3 , -122 , 3.2])
tf.math.abs(x_abs)
```

```python
x_abs_complex = tf.constant([2 + 1.1j])
tf.math.abs(x_abs_complex)
```

```python
x_sqt = tf.constant([4 , 16 , 5] , dtype='float32')
tf.sqrt(x_sqt)
```

```python
## Adding two tensors

x_a = tf.constant = [[[1,2],[2,3]]]
x_b = tf.constant = [[[1,1],[33 ,3]]]
tf.math.add(x_a , x_b)
```

```python
## BRODCASTING

x_a = tf.constant = [[[1,2],[2,3]]]
x_b = tf.constant = [[[1]]]
tf.add(x_a , x_b)
```

```python
# Multiplication
import tensorflow as tf

x_1 = tf.constant([[5,6,7,8,9]], dtype=tf.float16)
x_2 = tf.constant([[7],[6] ,[7]], dtype=tf.float16)
print(x_1.shape)
print(x_2.shape)
tf.math.multiply(x_1 , x_2)
```

```python
#ArgMax - It gives the position of the maximum value

x_argmax = tf.constant([1,2,3,54,6,8])
print(tf.argmax(x_argmax))
```

```python
x_argmax = tf.constant([[2,20,40,56],
                       [2,4,6,1],
                       [6,3,8,1]])
tf.argmax(x_argmax , 0)
```

```python
## Power 
x = tf.constant([1,2,3,4,5])
y= tf.constant([2,2,2,2,2])
tf.pow(x , y)
```

```python
## Reduce sum - sum all the values of the tensor
tensor_two_d = tf.constant([[1,2,3],
                           [4,5,6],
                           [6,7,8]] , dtype='float32')
print(tf.reduce_sum(tensor_two_d))
```

```python
## Reduce max - max of all the nums
tensor_two_d = tf.constant([[1,2,3],
                           [4,5,6],
                           [6,7,8]])
print(tf.reduce_max(tensor_two_d))
```

```python
## Reduce max by axis - if 0 then sum by column and if 1 then sum by row
print(tf.reduce_sum(tensor_two_d,0))
print(tf.reduce_sum(tensor_two_d,1))
```

```python
## Reduce mean
print(tf.math.reduce_mean(tensor_two_d,0))
```

```python
## Std deviation
print(tf.math.reduce_std(tensor_two_d , 0))
```

```python
## Top K - the later part twos are just the index of the top value

tf.math.top_k(tensor_two_d, k=1)
```

## Linear Algebra

## Matrix multiplication

```python
import tensorflow as tf
```

```python
x_1 = tf.constant([[1,2,3],
                  [-3,5,-1]])
x_2 = tf.constant([[1,2,0],
                  [3,5,-1],
                  [1,2,3]])

print(x_1.shape , x_2.shape)
tf.linalg.matmul( 
    x_1,
    x_2,
    transpose_a=False,
    transpose_b=False,
    adjoint_a=False,
    adjoint_b=False,
    a_is_sparse=False,
    b_is_sparse=False,
    output_type=None,
    grad_a=False,
    grad_b=False,
    name=None,
)
```

```python
#Matrix multiplication in short
print(x_1 @ x_2)
```

```python
## Transpositioning the matrix
print(x_1.shape)
x_1 = tf.transpose(x_1)
print(x_1.shape)
print(x_1)
```

```python
## Matrix multiplication of 3d Tensors

x_1 = tf.constant([[[1,2,3],
                    [3,5,1]],

                   [[3,4,2],
                   [4,4,1]],

                   [[1,1,1],
                   [2,2,2]]])

x_2 = tf.constant([[[4,2,5],
                    [1,7,9]],

                   [[3,4,9],
                   [6,7,1]],

                   [[1,1,8],
                   [2,8,2]]])

print(x_2 @ x_1)
```

```python
x_1 = tf.constant([[[1,2,3],
                    [3,5,1]],

                   [[3,4,2],
                   [4,4,1]],

                   [[1,1,1],
                   [2,2,2]]])

x_2 = tf.constant([
                    [[4,2],
                    [1,7],
                   [2,3]],

                   [[3,4],
                   [6,7],
                   [2,4]],

                   [[1,1],
                   [2,8],
                   [4,3]
                   ]])

print(x_1.shape , x_2.shape)
print(x_1 @ x_2)
```

```python
## Sparse matrixes

x_1 = tf.constant([[1,2,3],
                  [2,3,4]])
x_2 = tf.constant([[1,0],
                  [0,0],
                  [0,4]])
print(tf.linalg.matmul(x_1 , x_2 , b_is_sparse=True))
```

```python
## Adjoint matrix
print(x_1.shape)
print(tf.linalg.adjoint(x_1))
```

```python
print(tf.linalg.adjoint([[1,2,3],
                         [2,3,4],
                         [5,6,7]]))
```
